### Q10: Which category has the highest number of records?

Nan values: np.nan, -1, -2, -ve, "UNKNOWN", "-", "_", "missing", "?"

```python
def get_category(price):
    if price < 100000:
        return "Category 1"
    elif price < 200000:
        return "Category 2"
    elif 200000 <= price < 300000:
        return "Category 3"
    elif 300000 <= price < 400000:
        return "Category 4"
    return "Category 5"
```

```python
df["Category"] = df["SalePrice"].apply(get_category)
```

### Q12: How many features are there after performing the above transformation?

```python
from sklearn.preprocessing import MinMaxScaler, OneHotEncoder
from sklearn.impute import SimpleImputer
from sklearn.pipeline import Pipeline
from sklearn.compose import ColumnTransformer
```

11. LotFrontage: Impute with median then MinMax scaling
12. OverallQual: Impute with mean then MinMax scaling
13. CentralAir: Impute with most frequent value then one hot encode.

```python
X_train.shape
```

```python
transformer = ColumnTransformer([
    ("minmaxscaler", MinMaxScaler(), ["MSSubClass", "LotArea"]),
    ("LotFrontage", Pipeline([
        ("imputer", SimpleImputer(strategy="median")),
        ("minmaxscaler", MinMaxScaler())
    ]), ["LotFrontage"]),
    ("OverallQual", Pipeline([
        ("imputer", SimpleImputer(strategy="mean")),
        ("minmaxscaler", MinMaxScaler())
    ]), ["OverallQual"]),
    ("CentralAir", Pipeline([
        ("imputer", SimpleImputer(strategy="most_frequent")),
        ("encoder", OneHotEncoder())
    ]), ["CentralAir"])
], remainder="passthrough")
transformer
```

Use `GridSearchCV` for tuning the hyperparameters of the created pipeline on the training dataset.
- Values of `n_components` for PCA to be `[0.9, 0.95]`
- `lasso alpha` value to be taken as: `[10, 1, 0.01, 0.001]`
- scoring: `neg_mean_absolute_error`.
- Use 5 fold cross-validation.
- `n_jobs` = -1 (negative one) [it helps in using all the computational power to run this job]
> **Note**: Kindly ignore the warning.

```python
from sklearn.decomposition import PCA
from sklearn.linear_model import Lasso
```

```python
model_pipeline = Pipeline([
    ("transformer", PCA()),
    ("estimator", Lasso())
])
model_pipeline
```

```python
params = {
    "transformer__n_components": [0.9, 0.95],
    "estimator__alpha": [10, 1, 0.01, 0.001],
}
```

```python
cv_model = GridSearchCV(model_pipeline, params, scoring="neg_mean_absolute_error", cv=5, n_jobs=-1)
cv_model.fit(X_train, y_train)
```

### Q6: If we fit the pipeline on the training dataset, what will be the R2 score on the test dataset?

```python
r2_score(y_test, cv_model.predict(X_test))
```
