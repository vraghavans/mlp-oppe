```python
import numpy as np
import pandas as pd

from sklearn.metrics import mean_squared_error
from sklearn.pipeline import Pipeline
from sklearn.decomposition import PCA
from sklearn.preprocessing import PolynomialFeatures
from sklearn.feature_selection import SelectKBest, f_regression
from sklearn.model_selection import cross_val_score, GridSearchCV
from sklearn.linear_model import LinearRegression, Ridge, Lasso, SGDRegressor, RidgeCV
```

### Q1: Which dataset are you using for this Module 2?

- [x] Model_Building_1
- [ ] Model_Building_2
- [ ] Model_Building_3

```python
data = pd.read_csv("/kaggle/input/mlp-oppe1-mock-2/Model_Building_1.csv")
data.head()
```

```python
X = data.drop(columns="Sale Price")
y = data["Sale Price"]

X.head()
```

```python
X.shape
```

```python
X_train = X.iloc[:8271]
X_test = X.iloc[8272:]

y_train = y.iloc[:8271]
y_test = y.iloc[8272:]

X_train.shape, X_test.shape
```

### Q2: Apply LinearRegression on the train dataset(X_train and y_train). What is the R2 score on the test dataset(X_test and y_test). ( Upto 4 digits after decimal points)

```python
model = LinearRegression()
model.fit(X_train, y_train)
```

```python
round(model.score(X_test, y_test), 4)
```

### Q3: Enter the maximum value of 𝑅2 score you got using the LinearRegression model when computed with 5 folds from the training dataset (X_train and y_train) using cross_val_score. ( Upto 4 digits after decimal points) obtained.

```python
cv = cross_val_score(LinearRegression(), X_train, y_train, scoring="r2", cv=5)
cv
```

```python
round(cv.max(), 4)
```

### Q4: Train Ridge and Lasso with random_state=27 and keep other parameters as default using train dataset. Which one has the least "mean squared error" for the test dataset.
- [x] Ridge
- [ ] Lasso
- [ ] Both are Equal

```python
ridge_model = Ridge(random_state=27)
lasso_model = Lasso(random_state=27)

ridge_model.fit(X_train, y_train)
lasso_model.fit(X_train, y_train)
```

```python
y_ridge_preds = ridge_model.predict(X_test)
y_lasso_preds = lasso_model.predict(X_test)

ridge_mse = mean_squared_error(y_test, y_ridge_preds)
lasso_mse = mean_squared_error(y_test, y_lasso_preds)

mse_scores = pd.Series([ridge_mse, lasso_mse])
mse_scores.min(), mse_scores.argmin()
```

(Common Instructions for Question 14 and 15)

create a pipeline of the `PolynomialFeatures(interaction_only=True)` as transformer and Lasso as an estimator.

Use GridSearchCV for tuning the hyperparameters of the created pipeline on training dataset.

- Keep polynomial degree as : `[1,2]`
- lasso alpha value to be taken as : `[10,100,1000,10000]`
- scoring : `neg_mean_absolute_error`.
- cv = `5`
- n_jobs = `-1` (negative one) [it helps in using all the computational power to run this job]

(Note: Kindly ignore the warning.)

```python
pipeline = Pipeline([
    ("poly_trans", PolynomialFeatures(interaction_only=True)),
    ("estimator", Lasso())
])

pipeline
```

```python
params = {
    "poly_trans__degree": [1, 2],
    "estimator__alpha": [10, 100, 1000, 10000],
}
cv_model = GridSearchCV(pipeline, params, scoring="neg_mean_absolute_error", cv=5, n_jobs=-1)
cv_model.fit(X_train, y_train)
```

### Q6: Mark the best alpha value you got using above instructions.
- [ ] 10
- [ ] 100
- [x] 1000
- [ ] 10000


```python
cv_model.best_params_["estimator__alpha"]
```

### Q7: Enter the best polynomial degree value you got using above instructions.

```python
cv_model.best_params_["poly_trans__degree"]
```

(Common Instructions for Question 16 and 17)

To Reduce the number of dimensions of the training dataset with PCA. Fit the PCA model using following parameter values on training data.

- `n_components= 13`
- `svd_solver='full'`
- `whiten=True`
- `random_state=27`

```python
pca = PCA(n_components=13, svd_solver="full", whiten=True, random_state=27)
pca.fit(X_train, y_train)
```

### Q8: What is the sum of explained_variance_ratio_ ? (upto two digits after the decimal)

```python
round(pca.explained_variance_ratio_.sum(), 2)
```

### Q9: Use PCA transformed training feature matrix from earlier question and training labels to fit the RidgeCV estimator with cv=5 and having alpha values as [0.1,1,10,100,1000,10000]. Calculate the R2 score you got from the model for PCA transformed test dataset. (upto 4 digits after the decimal)

```python
X_train_trans = pca.transform(X_train)
X_test_trans = pca.transform(X_test)
```

```python
ridgecv_model = RidgeCV(alphas=[0.1, 1, 10, 100, 1000, 10000], cv=5)
ridgecv_model.fit(X_train_trans, y_train)
```

```python
round(ridgecv_model.score(X_test_trans, y_test), 4)
```

(Common Instructions for Question 18-20)

Construct a pipeline to apply following steps

- STEP-1 Selecting 10 best features based on the f_regression score function.
- STEP-2 Apply Lasso algorithm with random_state=27 on that

```python
lasso_pipeline = Pipeline([
    ("feature_selection", SelectKBest(f_regression, k=10)),
    ("model", Lasso(random_state=27))
])

lasso_pipeline
```

```python
lasso_pipeline.fit(X_train, y_train)
```

### Q10: What is the max P-VALUE you got for any feature of the train dataset?(upto 4 digits after the decimal)

```python
round(lasso_pipeline.named_steps["feature_selection"].pvalues_.max(), 4)
```

### Q11: What is the max F_STATISTICS score you got for any feature of the train dataset?  (upto 2 digits after the decimal)

```python
round(lasso_pipeline.named_steps["feature_selection"].scores_.max(), 2)
```

### Q12: What is the R2 score you got for this model using the test dataset? (upto 4 digits after the decimal)

```python
round(lasso_pipeline.score(X_test, y_test), 4)
```
